```python
# Create a new column 'Group' based on 'PercentSalaryHike'
# Ensure Treatment is correctly defined
hrdata["Treatment"] = (hrdata["PercentSalaryHike"] >= 18).astype(int)

# Ensure Post is correctly defined (Example: Based on YearsSinceLastPromotion)
hrdata["Post"] = (hrdata["YearsSinceLastPromotion"] >= 2).astype(int)

# Ensure Interaction Term is created correctly
hrdata["Treatment_Post"] = hrdata["Treatment"] * hrdata["Post"]
```

The top 5 most important variables to control for in your Difference-in-Differences (DiD) analysis are:

1. PerformanceRating – Employees with higher performance may receive larger salary hikes and be less likely to leave.
2. JobSatisfaction – Dissatisfied employees may leave regardless of salary hikes.
3. WorkLifeBalance – Poor balance can drive attrition, even if salaries increase.
4. YearsAtCompany – Long-tenured employees may be more stable, independent of salary hikes.
5. StockOptionLevel – Employees with stock options have incentives to stay longer.

# BASIC LOGISTIC REGRESSION

```python
import statsmodels.formula.api as smf

model = smf.logit("Attrition ~ PercentSalaryHike + JobSatisfaction", data=hrdata).fit()

print(model.summary())
```

# DiD

```python
import statsmodels.formula.api as smf

model = smf.logit(
    "Attrition ~ Treatment + Post + Treatment_Post + JobSatisfaction",
    data=hrdata
).fit()

print(model.summary())

```

#### Interpretation of the Difference-in-Differences Effect

**Treatment_Post is NOT statistically significant (p = 0.650)**
- This means salary hikes did not significantly reduce attrition in the treated group compared to the control group.
- The effect size is negative (-0.1644), suggesting a small reduction in attrition, but the p-value is too high to be considered meaningful.

**Key Takeaways**
1. Salary Hikes Did Not Causally Reduce Attrition
- The Treatment_Post coefficient is not significant, meaning there is no strong evidence that salary hikes led to lower attrition.
- Raises might not be the main reason employees stay or leave.

Job Satisfaction & Work-Life Balance Matter More Than Salary Hikes

Job Satisfaction (p = 0.000) → Higher satisfaction = much lower attrition.
Work-Life Balance (p = 0.018) → Employees with better work-life balance are less likely to leave.
3️⃣ Stock Options & Job Level Strongly Reduce Attrition

Employees at higher job levels or with stock options are far less likely to leave, suggesting long-term retention incentives are more effective than salary hikes alone.
4️⃣ Attrition Increased Slightly in the Post Period

The Post variable (p = 0.087) suggests that attrition increased slightly after salary hikes.
This could mean other external factors (e.g., job market conditions, company culture) played a role.

```python
import matplotlib.pyplot as plt
df_grouped = hrdata.groupby(['Post', 'Treatment'])['Attrition'].mean().unstack()
df_grouped.plot(marker='o')
plt.axvline(x=1, color='red', linestyle='--', label="Salary Hike Period")
plt.title("Attrition Rate Over Time (Treatment vs. Control)")
plt.xlabel("Period (Pre vs. Post)")
plt.ylabel("Attrition Rate")
plt.legend(["Control", "Treatment", "Salary Hike Period"])
plt.show()
```

#### Interpretation of the Attrition Rate Over Time (DiD Plot)
1. Key Observations from the Graph
**Control Group (Blue Line)**
- Attrition decreased over time.
- Suggests external factors (not salary hikes) may have contributed to lower attrition.

**Treatment Group (Orange Line)**
- Attrition also decreased after the salary hike.
- The drop is slightly larger than in the control group.
- BUT the statistical model showed this difference was not significant (p = 0.650 in logistic regression).

**Salary Hike Period (Red Dashed Line)**
- The salary hike was implemented at x = 1.0 (Post period).
- After the hike, attrition continued to decline, but there’s no strong evidence this was due to the raises.

#### What Does This Mean for the Business?
1. Salary hikes may not be the main driver of retention.
- The drop in attrition happens in both groups.
- External factors (job market, leadership, culture, etc.) might be more influential than salary hikes.

2. Raises might work, but only for certain groups.
- We should test interactions with job roles, departments, and performance ratings.
- If raises only reduce attrition for certain employees, a targeted salary strategy may be better than company-wide increases.

3. More research is needed on retention strategies.
- Given Work-Life Balance and Job Satisfaction were significant predictors, improving these may be more effective than salary hikes alone.
